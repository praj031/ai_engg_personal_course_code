Python supports Object-Oriented Programming (OOP) without requiring a `new` keyword for instantiation[cite: 1].

### Core Concepts
* `__init__`: Constructor method called when initializing an instance[cite: 1].

> **Resource:** Tutorial on [Python Classes and Objects](https://www.w3schools.com/python/python_classes.asp)[cite: 1].

````python
class Dog:
    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self, colour : str) -> str:
        return f"{self.name} says Wooff!!"


leo = Dog("Leo", "Labra")
raka = Dog("Raka", "Pomerian")

print(f"Dog Name : {leo.name}   &     Dog Breed : {leo.breed}")
print(f"Dog Name : {raka.name}   &     Dog Breed : {raka.breed}")
print(f"Dog Name : {leo.name} & {leo.bark("brown")}")
````

@classmethod is best understood from Java by starting with what it is: a decorator — syntax wrapping the method definition to change how Python treats it. The line above def is doing something, not labeling it.
Without the decorator, an ordinary method receives the instance as its first argument (self). With @classmethod, the method receives the class itself as its first argument, conventionally named cls

Class method is used to consturct a different object in the way we want to create it. 

````python
class User:

    def __init__(self,name):
        self.name = name

    @classmethod
    def from_dist(cls,data):
        return cls(data["name"])

#Here you can see there is no use of the object creation and things can be called along. 
print(f" The value from {User.from_dist({"name":"Pritish"})}")
````

@static method : Think of it as a function inside a class namespace.

We can do some util operation using this. or we can use it make some constant changes in any method. Also we can do it without having to create the object, we can direclty use it.

````python
class Dog:
    species = "Dog"

    @staticmethod
    def run():
        print("Dog is running")

    @staticmethod
    def add(a,b):
        return a+b

Dog.run()

Dog.add(6,9)

print(f"The sum = {Dog.add(5,4)}")
````

@Property

Used to make a method readable like a plain attribute -- loosely likea a java getter but called without perentheses. You will see this in libraries that expose computed values without breaking the external object.attribute access pattern.

It's like defining dynamic properties of a class.

Usecase : IF we want to create a dynamic variable, suppose the ask is to use one variable that holdes the values of multiple variables. So here we define it as the property.

````python
class User:
    def __init__(self,first,last):
        self.first = first
        self.last = last

    @property
    def final_name(self):
        return f"{self.first}{self.last}" 


value = User.final_name

print(value)
````

## 2. Inheritance and `super()`

Subclasses inherit attributes and methods from base classes[cite: 1]. Use `super()` to delegate initialization or invocation to parent classes[cite: 1].

> **Resource:** Read more about [Python Inheritance](https://www.w3schools.com/python/python_inheritance.asp).

````python
class Animal:

    def speak(self):
        print("Some sount")

class Dog(Animal):
    #Redefining the method from the parents class
    #more like overridding
    def speak(self):
        print("woof");        


dog = Dog()
dog.speak()
````

````python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def speak(self) -> str:
        return "..."

class Dog(Animal):
    def __init__(self, name: str, breed: str):
        super().__init__(name)           # Delegate initialization to Animal
        self.breed = breed

    def speak(self) -> str:              # Override method
        return "Woof!"

generic_animal = Animal("Some Animal")
rex = Dog("Rex", "Labrador")

print(f"generic_animal.speak() -> {generic_animal.speak()}")
print(f"rex.speak()            -> {rex.speak()}")
print(f"rex.name (from parent __init__) -> {rex.name}")
print(f"rex.breed (from Dog's own __init__) -> {rex.breed}")
````

# Duck Typing

Python often cares about **behavior** rather than an explicit inheritance relationship.
Python's "care about behavior, not pedigree" rule for how functions accept objects. The name comes from the saying: "If it walks like a duck and quacks like a duck, it's a duck." Python applies that to code: if an object has a speak() method, it can be passed to make_sound(), and Python doesn't care what class it came from.

````python
class Dog:
    def speak(self):
        return "Woof"


class Cat:
    def speak(self):
        return "Meow"


def make_sound(animal): # aminal is the parameter which help us to call any method
    return animal.speak()

print(make_sound(Dog()))
print(make_sound(Cat()))
````

Dataclass help in python to reduce the boiler plate codes. 

@dataclass, the closest Python has to Java records/Lombok. The decorator generates __init__ (plus a comparison method and more) from the annotated fields — and here, unlike bare annotations in normal classes, the annotations are used because @dataclass reads them:

````python
from dataclasses import dataclass

@dataclass
class Animal:
    name: str
    breed: str


final = Animal("Dog", "Labra")   # generated __init__ takes both args
print(final.name)                # Dog
````

## 2. Variadic Arguments: `*args` and `**kwargs`

Variadic parameters allow functions to accept an arbitrary number of arguments[cite: 1]. This design pattern is used across AI frameworks (e.g., PyTorch, LangChain, Hugging Face) for passing optional model configurations[cite: 1].

* `*args`: Collects positional arguments into a **tuple**[cite: 1].
* `**kwargs`: Collects keyword arguments into a **dictionary**[cite: 1].

[args and kwargs](https://www.geeksforgeeks.org/python/args-kwargs-python/)[cite: 1]

The use case is like, when we don't have a fix use case of adding in the arguments. Then we use this type of var argument.

````python
#Example of the passing of only the arguments. Taking in the linear list of something

def add_number(*arg):
    print(f"The type of the agrumet is {type(arg)}")
    print(f"The values to be passed inside the argument {arg}")

add_number("Pritish","Anuj","Likhita","Arun")    
````

````python
#Example of the passing of the keywords args. When we proivde a dictionary 

def print_profile(**kwarg):
    print(kwarg)

value = print_profile(name="Anuj",age=30,isActive=True)
print(f"The final output = {value}")
````

````python
def create_profile(name, role, experience):
    print(name)
    print(role)
    print(experience)

profile = {
    "name": "Shivam",
    "role": "SDE",
    "experience": 14
}

# profile = ["Shivam", "SDE", 14]

create_profile(name="Anuj", role="SDE", experience=5)
create_profile(**profile)  

##This is un-packing the values in any format.
````

## 3. Dynamic Instance Attributes & Lifecycle

### Homework: AI Agent Session

Create an `AgentSession` class with a constructor that accepts:

```python
user_id
```

Initially, the object should only contain `user_id`.

Then dynamically add these attributes after creating the object:

```python
model
message_count
is_active
```

Add methods:

```python
start()
end()
```

`start()` should activate the session and `end()` should deactivate it.

Also investigate and demonstrate what happens when an object is deleted using:

```python
del session
```

Use `__del__()` to observe when the object's lifecycle ends.

---

````python
class AgentSession:
    def __init__(self, user_id):
        self.user_id = user_id

    def start(self):
        self.is_active = True
        print(f"Session for {self.user_id} started.")

    def end(self):
        self.is_active = False
        print(f"Session for {self.user_id} ended.")

    def __del__(self):
        print(f"AgentSession for {self.user_id} is being destroyed.")


session = AgentSession("Pritish")

print("Initial attributes:", vars(session))

# Dynamically add attributes after object creation
session.model = "gpt-5"
session.message_count = 12
session.is_active = True

print("Attributes after dynamic additions:", vars(session))

session.start()
session.end()

print("Final attributes:", vars(session))

del session
````
